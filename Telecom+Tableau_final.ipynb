# Анализ текущего уровня потребительской лояльности



## Описание проекта

**Заказчик:** крупная телекоммуникационная компания, предоставляющая свои услуги в странах СНГ.

**Задача:** определить текущий уровень потребительской лояльности (NPS - Net Promoter Score) среди клиентов из России.

**Источник данных:** для определения уровня лояльности компания провела опрос среди своих клиентов. Клиентам задавали классический вопрос: «Оцените по шкале от 1 до 10 вероятность того, что вы порекомендуете компанию друзьям и знакомым».

**Методика расчета NPS:** Чтобы оценить результаты опроса, оценки обычно делят на три группы:

9-10 баллов — «cторонники» (promoters);

7-8 баллов — «нейтралы» (passives);

0-6 баллов — «критики» (detractors).

Итоговое значение NPS рассчитывается по формуле: % «сторонников» - % «критиков».

### Описание данных

Данные представлены следующими таблицами:

* `user` - содержит основную информацию о клиентах и первичные ключи к остальным таблицам;

* `location` - справочник территорий, в которых телеком-компания оказывает услуги;

* `age_segment` - данные о возрастных сегментах клиентов;

* `traffic_segment` - данные о выделяемых сегментах по объёму потребляемого трафика;

* `lifetime_segment` - данные о выделяемых сегментах по количеству месяцев «жизни» клиента (как давно пользователь является клиентом компании).

### Выгрузка данных

Для дальнейшей обработки данных и постороения дашбордов необходимо получить таблицу со следующими параметрами:

| Parameter         | Description                                                                   | 
|-------------------|-------------------------------------------------------------------------------|
| user_id	        | Идентификатор клиента                                                         | 
| lt_day	        | Количество дней «жизни» клиента                                               | 
| is_new	        | Поле хранит информацию о том, является ли клиент новым                        |
| age	            | Возраст                                                                       |
| gender_segment	| Пол (для удобства работы с полем преобразуйте значения в текстовый вид)       |
| os_name	        | Тип операционной системы                                                      |
| cpe_type_name	    | Тип устройства                                                                | 
| country	        | Страна проживания                                                             | 
| city	            | Город проживания                                                              |
| age_segment	    | Возрастной сегмент                                                            |
| traffic_segment    | Сегмент по объёму потребляемого трафика                                       |
| lifetime_segment  | Сегмент по количеству дней «жизни»                                            | 
| nps_score	        | Оценка клиента в NPS-опросе                                                   |
| nps_group	        | Поле хранит информацию о том, к какой группе относится оценка клиента в опросе| 

```python
query = """
SELECT u.user_id,
u.lt_day,
    CASE
        WHEN u.lt_day <= 365 THEN True
        ELSE False
    END as is_new,
u.age,
    CASE
        WHEN u.gender_segment = 1 THEN 'women'
        WHEN u.gender_segment = 0 THEN 'men'
        ELSE 'nd'
    END as gender_segment,
u.os_name,
u.cpe_type_name,
l.country,
l.city,
a.title as age_segment,
t.title as traffic_segment,
f.title as lifetime_segment,
u.nps_score,
    CASE
        WHEN u.nps_score <= 6 THEN 'detractors'
        WHEN u.nps_score > 6 AND u.nps_score <= 8 THEN 'passives'
        ELSE 'promoters'
    END as nps_group
FROM user as u JOIN location as l ON u.location_id = l.location_id
    JOIN age_segment as a ON u.age_gr_id = a.age_gr_id
    JOIN traffic_segment as t ON u.tr_gr_id = t.tr_gr_id
    JOIN lifetime_segment as f ON u.lt_gr_id = f.lt_gr_id;
"""
```

```python
dfj = pd.read_sql(query, engine)
dfj.head(10)
```
